```python
import os
from openai import OpenAI
from dotenv import load_dotenv
```

```python
load_dotenv(override=True)
google_api_key = os.getenv("GOOGLE_API_KEY")
gemini_base_url = os.getenv("GEMINI_BASE_URL")
```

```python
messages = [{"role": "user", "content": "What is 2+2?"}]
```

```python
gemini = OpenAI(base_url=gemini_base_url, api_key=google_api_key)
response = gemini.chat.completions.create(
    model="gemini-2.5-flash", 
    messages=messages
)

print(response.choices[0].message.content)
```

```python
question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
response = gemini.chat.completions.create(
    model="gemini-2.5-flash",
    messages=messages
)

question = response.choices[0].message.content

print(question)
```

```python
messages = [{"role": "user", "content": question}]
```

```python
response = gemini.chat.completions.create(
    model="gemini-2.5-flash",
    messages=messages
)

answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
# First create the messages:

messages = [{"role": "user", "content": "Selecciona un sector de la economia que te interese"}]

# Then make the first call:

response = gemini.chat.completions.create(
    model="gemini-2.5-flash",
    messages=messages
)

# Then read the business idea:

business_idea = response.choices[0].message.content
print(business_idea)

# And repeat! In the next message, include the business idea within the message
```

```python
messages = [{"role": "user", "content": "Presenta un problema en la industria de " + business_idea + ": algo desafiante que podría ser propicio para una solución Agentic."}]

response = gemini.chat.completions.create(
    model="gemini-2.5-flash",
    messages=messages
)

pain_point = response.choices[0].message.content
print(pain_point)

```

```python
messages = [{"role": "user", "content": "Presenta 3 soluciones agenticas para el problema de " + pain_point + " en la industria de " + business_idea + "."}]

response = gemini.chat.completions.create(
    model="gemini-2.5-flash",
    messages=messages
)

solutions = response.choices[0].message.content
print(solutions)
```

```python
from IPython.display import Markdown, display

display(Markdown(solutions))
```
